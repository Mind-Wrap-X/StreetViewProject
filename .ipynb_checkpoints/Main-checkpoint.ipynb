```python
# API Parameters
size = "640x640"
pitch = 0
fov = 120
headings = [0, 90, 180, 270]
```

```python
# Loop through each location and heading
for i, (longitude, latitude) in enumerate(road_coordinates):
    for heading in headings:
         url = f"https://maps.googleapis.com/maps/api/streetview?size={size}&location={latitude},{longitude}&heading={heading}&pitch={pitch}&fov={fov}&key={API_KEY}&return_error_code=true"

         try:
            # Make the request with a timeout to prevent indefinite waiting
            response = requests.get(url, timeout=10)

            # Raise an HTTPError for bad responses (4xx or 5xx)
            response.raise_for_status()

            # Create a descriptive filename
            filename = f"{ROAD_NAME}_{START_INTERSECTION}_{END_INTERSECTION}_{i}_{heading}.jpg"
            filepath = os.path.join(OUTPUT_FOLDER, filename)

            # Save the image content
            with open(filepath, "wb") as f:
                f.write(response.content)

            print(f"Successfully downloaded: {filename}")

         except requests.exceptions.HTTPError as err:
            # This handles HTTP errors (e.g., 404, 500)
            print(f"HTTP Error for {latitude},{longitude} (heading {heading}): {err}")
         except requests.exceptions.RequestException as err:
            # This handles all other request-related errors (e.g., connection issues, timeouts)
            print(f"Request Error for {latitude},{longitude} (heading {heading}): {err}")
         except Exception as err:
            # A catch-all for any other unexpected errors
            print(f"An unexpected error occurred for {latitude},{longitude} (heading {heading}): {err}")
    # Optional: Add a small delay between requests to be respectful of the API's rate limits
    time.sleep(0.5)

print("Done!")
print("All done!")
```
