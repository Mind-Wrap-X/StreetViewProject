```python

# Define the road and intersections
ROAD_NAME = "E_California_Ave"
START_INTERSECTION = "N_Sunnyvale_Ave"
END_INTERSECTION = "Fair_Oaks_Ave"

OUTPUT_FOLDER = f"data/street_view_images/{ROAD_NAME}_{START_INTERSECTION}_{END_INTERSECTION}"


if not os.path.exists(OUTPUT_FOLDER):
    os.makedirs(OUTPUT_FOLDER) # Create the folder if it does not exist.

road_coordinates = [
    (37.381191, -122.027311),
    (37.381200, -122.026809),
    (37.381191, -122.026467),
    (37.381139, -122.026296),
    (37.381111, -122.026125),
    (37.381074, -122.025936),
    (37.381022, -122.025718),
    (37.380999, -122.025494),
    (37.380952, -122.025305),
    (37.380881, -122.025004),
    (37.380816, -122.024579),
    (37.380732, -122.024292),
    (37.380584, -122.023474),
    (37.380456, -122.022905),
    (37.380328, -122.022347),
    (37.380251, -122.021993),
    (37.380123, -122.021381),
    (37.379996, -122.020813),
    (37.379927, -122.020384),
    (37.379808, -122.019955),
    (37.379808, -122.019955),
    (37.379638, -122.019150)
]

'''road_coordinates = [
    (37.381200, -122.027228),
]'''
```

```python
# API Parameters
size = "640x640"
pitch = -6
fov = 50
headings = [0, 90, 180, 270, 103.47]
```

```python
# Loop through each location and heading
for i, (longitude, latitude) in enumerate(road_coordinates):
    for heading in headings:
         url = f"https://maps.googleapis.com/maps/api/streetview?size={size}&location={longitude},{latitude}&heading={heading}&pitch={pitch}&fov={fov}&key={API_KEY}&return_error_code=true"

         try:
            # Make the request with a timeout to prevent indefinite waiting
            response = requests.get(url, timeout=10)

            # Raise an HTTPError for bad responses (4xx or 5xx)
            response.raise_for_status()

            # Create a descriptive filename
            filename = f"{ROAD_NAME}_{START_INTERSECTION}_{END_INTERSECTION}_{i}_{heading}_{fov}.jpg"
            filepath = os.path.join(OUTPUT_FOLDER, filename)

            # Save the image content
            with open(filepath, "wb") as f:
                f.write(response.content)

            print(f"Successfully downloaded: {filename}")

         except requests.exceptions.HTTPError as err:
            # This handles HTTP errors (e.g., 404, 500)
            print(f"HTTP Error for {latitude},{longitude} (heading {heading}): {err}")
         except requests.exceptions.RequestException as err:
            # This handles all other request-related errors (e.g., connection issues, timeouts)
            print(f"Request Error for {latitude},{longitude} (heading {heading}): {err}")
         except Exception as err:
            # A catch-all for any other unexpected errors
            print(f"An unexpected error occurred for {latitude},{longitude} (heading {heading}): {err}")
    # Optional: Add a small delay between requests to be respectful of the API's rate limits
    time.sleep(0.5)

print("Done!")
print("All done!")
```

```python
import math

def get_bearing(start_coord, end_coord):
    """
    Calculates the compass bearing between two coordinates.

    Args:
        start_coord (tuple): A tuple (latitude, longitude) for the starting point.
        end_coord (tuple): A tuple (latitude, longitude) for the ending point.

    Returns:
        float: The bearing in degrees, from 0 to 360, where 0 is North.
    """
    lat1 = math.radians(start_coord[0])
    lon1 = math.radians(start_coord[1])
    lat2 = math.radians(end_coord[0])
    lon2 = math.radians(end_coord[1])

    delta_lon = lon2 - lon1

    # Calculate the components for the bearing
    x = math.sin(delta_lon) * math.cos(lat2)
    y = math.cos(lat1) * math.sin(lat2) - (math.sin(lat1) * math.cos(lat2) * math.cos(delta_lon))

    # Calculate the initial bearing in radians and convert to degrees
    initial_bearing = math.atan2(x, y)
    initial_bearing = math.degrees(initial_bearing)

    # Normalize the bearing to a 0-360 degree compass
    compass_bearing = (initial_bearing + 360) % 360

    return compass_bearing

# Example with your coordinates
first_coord = (37.381191, -122.027311)
last_coord = (37.379638, -122.019150)

angle = get_bearing(first_coord, last_coord)

print(f"The calculated heading from the first coordinate to the last is: {angle:.2f} degrees.")
```
